```python
import pandas as pd
import numpy as np

# Load IBM Falcon 9 dataset
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"

data = pd.read_csv(url)

# Show first rows
data.head()
```

```python
# Keep Falcon 9 only
data_falcon9 = data[data['BoosterVersion'] != 'Falcon 1'].copy()

# Reset flight numbers
data_falcon9.loc[:, 'FlightNumber'] = list(range(1, data_falcon9.shape[0] + 1))

data_falcon9.head()
```

```python
# Check missing values before
data_falcon9.isnull().sum()
```

```python
# Replace missing PayloadMass with the mean
mean_payload_mass = data_falcon9['PayloadMass'].mean()

data_falcon9['PayloadMass'] = data_falcon9['PayloadMass'].replace(np.nan, mean_payload_mass)

# Check missing values after
data_falcon9.isnull().sum()
```

```python
# Save final assignment file
data_falcon9.to_csv('dataset_part_1.csv', index=False)

print("Saved dataset_part_1.csv")
data_falcon9.head()
```
